```python
#data Visualization using matplotlib library
#1. How has startup funding trend changed over the years?
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

funding_by_year = data.groupby('year')['amount'].sum().reset_index()

plt.figure(figsize=(10, 6))
sns.lineplot(x='year', y='amount', data=funding_by_year, marker='o')
plt.title('Startup Funding Trend Over the Years')
plt.xlabel('Year')
plt.ylabel('Total Funding Amount')
plt.grid(True)
plt.xticks(funding_by_year['year'])
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

startups_by_city = data['city'].value_counts().reset_index()
startups_by_city.columns = ['city', 'count']

top_10_cities = startups_by_city.head(10)

plt.figure(figsize=(12, 7))
sns.barplot(x='city', y='count', data=top_10_cities,palette='viridis')
plt.title('Top 10 Cities by Number of Startups')
plt.xlabel('City')
plt.ylabel('Number of Startups')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y',linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
#3.Which startup sectors have the highest number of funded startup ? barchart horigentle
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

startups_by_vertical = data['vertical'].value_counts().reset_index()
startups_by_vertical.columns = ['vertical', 'count']

top_10_verticals = startups_by_vertical.head(10)

plt.figure(figsize=(10, 6))
sns.barplot(x='count', y='vertical', data=top_10_verticals)
plt.title('Top 10 Startup Verticals by Number of Startups')
plt.xlabel('Number of Startups')
plt.ylabel('Vertical')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
funding_rounds = data['round'].value_counts().head(7)


num_top_rounds = 7
if len(funding_rounds) > num_top_rounds:
    top_rounds = funding_rounds.head(num_top_rounds - 1)
    other_rounds_sum = funding_rounds.iloc[num_top_rounds - 1:].sum()
    funding_distribution = pd.concat([top_rounds, pd.Series({'Other': other_rounds_sum})])
else:
    funding_distribution = funding_rounds
plt.figure(figsize=(10, 6))
colors =["Black","blue","green","orange","pink","yellow","red"]
plt.pie(funding_distribution, labels=funding_distribution.index, autopct='%1.1f%%', startangle=140,colors=colors)
plt.title('Distribution of Funding Rounds')
plt.axis('equal')
plt.show()
```

```python
#5.how are startup funding amounts distributed? histogram
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.histplot(data['amount'], bins=50)
plt.title('Distribution of Startup Funding Amounts')
plt.xlabel('Funding Amount')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
